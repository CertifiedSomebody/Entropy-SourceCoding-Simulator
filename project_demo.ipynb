```python
# --- Cell 2: Text Example ---
from Entropy_InfoRate_Simulator import (
    text_to_symbols,
    entropy_of_sequence,
    build_huffman_code,
    build_shannon_fano_code,
    average_code_length,
    probs_from_sequence
)

# Input text
text = "this is an example of entropy and coding demonstration"
seq = list(text)  # convert to sequence of symbols

# Compute probabilities dictionary
prob_dict = probs_from_sequence(seq)  # dict: symbol -> probability

# Compute entropy
H_text = entropy_of_sequence(seq)
print("Text Example:")
print("Entropy (bits/symbol):", H_text)

# Huffman coding
huff_codes = build_huffman_code(prob_dict)  # dict: symbol -> code
L_huff = average_code_length(huff_codes, prob_dict)  # avg length
print("Huffman Avg Code Length:", L_huff)

# Shannon-Fano coding
sf_codes = build_shannon_fano_code(prob_dict)  # dict: symbol -> code
L_sf = average_code_length(sf_codes, prob_dict)  # avg length
print("Shannon-Fano Avg Code Length:", L_sf)
```

```python
from Entropy_InfoRate_Simulator import lz78_entropy_estimator

# Information rate (bits/sec)
H_est_bits_per_symbol = lz78_entropy_estimator(seq)  # bits per symbol
bitrate = 1000  # symbols per second
R_text = H_est_bits_per_symbol * bitrate  # bits per second
print("\nInformation Rate:", R_text, "bits/sec")
```
